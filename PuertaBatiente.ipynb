🚪 Cotizador de Puertas de Baño Batientes (Puerta + Panel Fijo)
Empresa: IlumiVidrios / Cristalería
Instrucciones:

Revisa las tablas de precios de la base de datos (Puerta, Panel Fijo, Esmerilado y Accesorios).

En "Datos del Proyecto", ingresa el Ancho Total y el Ancho de la Puerta. El panel fijo se calculará automáticamente.

Ejecute la celda final para obtener el desglose detallado.

```python
# ==============================================================================
# TABLA 1A: Precios por Rangos de Ancho para HOJA DE PUERTA (en metros)
# ==============================================================================
tabla_rangos_puerta = [
    {"min": 0.500, "max": 0.599, "precio": 80.00},
    {"min": 0.600, "max": 0.650, "precio": 90.00},
    {"min": 0.651, "max": 0.700, "precio": 95.00}
]

# ==============================================================================
# TABLA 1B: Precios por Rangos de Ancho para PANEL FIJO (en metros)
# ==============================================================================
tabla_rangos_fijo = [
    {"min": 0.000, "max": 0.100, "precio": 0.00},
    {"min": 0.101, "max": 0.239, "precio": 35.00},
    {"min": 0.240, "max": 0.300, "precio": 44.00},
    {"min": 0.301, "max": 0.500, "precio": 62.00},
    {"min": 0.501, "max": 0.700, "precio": 81.00},
    {"min": 0.701, "max": 0.900, "precio": 99.00}
]

# ==============================================================================
# TABLA 2: Opciones y Adicionales por Esmerilado de Cristales (USD)
# ==============================================================================
tabla_esmerilado = {
    "No": 0.00,
    "Si": 30.00
}

# ==============================================================================
# TABLA 3: Catálogo de Accesorios y Herrajes para Puerta Batiente (USD)
# ==============================================================================
tabla_accesorios = {
    "Tirador Tipo O": 17.00,
    "Tirador Tipo Negro": 20.00,
    "Tirador Pomo Doble": 8.00,
    "Tirador Pomo Negro": 10.00,
    "Barra Tensora": 20.00,
    "Tirador Toallero": 25.00
}

print("✅ Base de datos para Puertas Batientes cargada correctamente.")
```

```python
# --- DATOS DEL CLIENTE ---
nombre_cliente = "Eithan Silva"
referencia_proyecto = "Baño Principal"

# --- 1. MEDIDAS DEL PROYECTO (En metros) ---
ancho_total = 0.60   # Ancho total del vano
ancho_puerta = 0.60  # Ancho asignado a la puerta batiente

# --- 2. SELECCIÓN DE ESMERILADO ---
opcion_esmerilado_seleccionada = "No"

# --- 3. SELECCIÓN DE ACCESORIOS ---
accesorios_seleccionados = [
    "Tirador Toallero"
]
```

```python
# --- 1. CÁLCULO DEL PANEL FIJO ---
ancho_fijo = round(ancho_total - ancho_puerta, 3)

# Validar que la medida del fijo sea coherente
if ancho_fijo <= 0:
    # --- 2. EVALUAR TABLA 1A: Precio Puerta ---
    precio_puerta = 0.0
    rango_puerta_aplicado = "Fuera de rango"

    for rango in tabla_rangos_puerta:
        if rango["min"] <= round(ancho_puerta, 3) <= rango["max"]:
            precio_puerta = rango["precio"]+100
            rango_puerta_aplicado = f"{rango['min']:.3f} m a {rango['max']:.3f} m"
            break

    if precio_puerta == 0.0:
        print(f"⚠️ ADVERTENCIA: La puerta ({ancho_puerta:.3f}m) está fuera de rango en Tabla 1A.")

    # --- 3. EVALUAR TABLA 1B: Precio Panel Fijo ---
    precio_fijo = 0.0
    rango_fijo_aplicado = "Fuera de rango"

    for rango in tabla_rangos_fijo:
        if rango["min"] <= round(ancho_fijo, 3) <= rango["max"]:
            precio_fijo = rango["precio"]
            rango_fijo_aplicado = f"{rango['min']:.3f} m a {rango['max']:.3f} m"
            break

    if precio_fijo == 0.0:
        print(f"⚠️ NOTA: Esta puerta no llevara panel fijo.")

    # --- 4. EVALUAR TABLA 2: Esmerilado ---
    precio_tabla2 = tabla_esmerilado.get(opcion_esmerilado_seleccionada, 0.0)

    # --- 5. EVALUAR TABLA 3: Suma de Accesorios ---
    precio_tabla3 = 0.0
    desglose_accesorios = []

    for acc in accesorios_seleccionados:
        if acc in tabla_accesorios:
            costo_acc = tabla_accesorios[acc]
            precio_tabla3 += costo_acc
            desglose_accesorios.append((acc, costo_acc))

    # --- CÁLCULO TOTAL ---
    subtotal_vidrios = precio_puerta + precio_fijo
    total_cotizacion = subtotal_vidrios + precio_tabla2 + precio_tabla3

    # --- GENERACIÓN DEL RESUMEN DETALLADO ---
    print("=" * 60)
    print("📄 COTIZACIÓN DE PUERTA DE BAÑO BATIENTE + PANEL FIJO")
    print(f"👤 Cliente: {nombre_cliente} | 📍 Ubicación: {referencia_proyecto}")
    print("=" * 60)
    print(f"📏 Ancho Total Ingresado:   {ancho_total:.2f} m")
    print(f" ├── 🚪 Puerta Batiente:    {ancho_puerta:.2f} m [{rango_puerta_aplicado}] -> ${precio_puerta:.2f}")
    print(f" └── 🖼️ Panel Fijo:         {ancho_fijo:.2f} m [{rango_fijo_aplicado}] -> ${precio_fijo:.2f}")
    print(f"💵 Subtotal Cristales:     ${subtotal_vidrios:.2f}")
    print("-" * 60)
    print("🎨 Esmerilado (Tabla 2):")
    print(f"  • {opcion_esmerilado_seleccionada}: +${precio_tabla2:.2f}")
    print("-" * 60)
    print("🛠️ Accesorios Seleccionados (Tabla 3):")
    if desglose_accesorios:
        for nombre_acc, costo in desglose_accesorios:
            print(f"  • {nombre_acc}: +${costo:.2f}")
    else:
        print("  • Ningún accesorio adicional seleccionado.")
    print("=" * 60)
    print(f"💰 PRECIO TOTAL A COBRAR:  ${total_cotizacion:.2f}")
    print("=" * 60)
else:
    # --- 2. EVALUAR TABLA 1A: Precio Puerta ---
    precio_puerta = 0.0
    rango_puerta_aplicado = "Fuera de rango"

    for rango in tabla_rangos_puerta:
        if rango["min"] <= round(ancho_puerta, 3) <= rango["max"]:
            precio_puerta = rango["precio"]+50
            rango_puerta_aplicado = f"{rango['min']:.3f} m a {rango['max']:.3f} m"
            break

    if precio_puerta == 0.0:
        print(f"⚠️ ADVERTENCIA: La puerta ({ancho_puerta:.3f}m) está fuera de rango en Tabla 1A.")

    # --- 3. EVALUAR TABLA 1B: Precio Panel Fijo ---
    precio_fijo = 0.0
    rango_fijo_aplicado = "Fuera de rango"

    for rango in tabla_rangos_fijo:
        if rango["min"] <= round(ancho_fijo, 3) <= rango["max"]:
            precio_fijo = rango["precio"]+50
            rango_fijo_aplicado = f"{rango['min']:.3f} m a {rango['max']:.3f} m"
            break

    if precio_fijo == 0.0:
        print(f"⚠️ ADVERTENCIA: El panel fijo ({ancho_fijo:.3f}m) está fuera de rango en Tabla 1B.")

    # --- 4. EVALUAR TABLA 2: Esmerilado ---
    precio_tabla2 = tabla_esmerilado.get(opcion_esmerilado_seleccionada, 0.0)

    # --- 5. EVALUAR TABLA 3: Suma de Accesorios ---
    precio_tabla3 = 0.0
    desglose_accesorios = []

    for acc in accesorios_seleccionados:
        if acc in tabla_accesorios:
            costo_acc = tabla_accesorios[acc]
            precio_tabla3 += costo_acc
            desglose_accesorios.append((acc, costo_acc))

    # --- CÁLCULO TOTAL ---
    subtotal_vidrios = precio_puerta + precio_fijo
    total_cotizacion = subtotal_vidrios + precio_tabla2 + precio_tabla3

    # --- GENERACIÓN DEL RESUMEN DETALLADO ---
    print("=" * 60)
    print("📄 COTIZACIÓN DE PUERTA DE BAÑO BATIENTE + PANEL FIJO")
    print(f"👤 Cliente: {nombre_cliente} | 📍 Ubicación: {referencia_proyecto}")
    print("=" * 60)
    print(f"📏 Ancho Total Ingresado:   {ancho_total:.2f} m")
    print(f" ├── 🚪 Puerta Batiente:    {ancho_puerta:.2f} m [{rango_puerta_aplicado}] -> ${precio_puerta:.2f}")
    print(f" └── 🖼️ Panel Fijo:         {ancho_fijo:.2f} m [{rango_fijo_aplicado}] -> ${precio_fijo:.2f}")
    print(f"💵 Subtotal Cristales:     ${subtotal_vidrios:.2f}")
    print("-" * 60)
    print("🎨 Esmerilado (Tabla 2):")
    print(f"  • {opcion_esmerilado_seleccionada}: +${precio_tabla2:.2f}")
    print("-" * 60)
    print("🛠️ Accesorios Seleccionados (Tabla 3):")
    if desglose_accesorios:
        for nombre_acc, costo in desglose_accesorios:
            print(f"  • {nombre_acc}: +${costo:.2f}")
    else:
        print("  • Ningún accesorio adicional seleccionado.")
    print("=" * 60)
    print(f"💰 PRECIO TOTAL A COBRAR:  ${total_cotizacion:.2f}")
    print("=" * 60)
```
