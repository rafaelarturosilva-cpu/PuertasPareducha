# 🚿 Cotizador de Puertas de Baño Corredizas
**Empresa:** IlumiVidrios / Cristalería
**Instrucciones:**
1. Revisa las 3 tablas de la base de datos y ajusta los valores si tus precios cambian.
2. Ingresa las dimensiones y selecciones del cliente en la sección de "Datos del Proyecto".
3. Ejecuta la celda final para obtener el desglose y costo total.

```python
# ==============================================================================
# TABLA 1: Precios por Rangos de Ancho (en metros)
# ==============================================================================
# Puedes modificar o agregar cuantos rangos necesites
tabla_rangos_ancho = [
    {"min": 0.75, "max": 0.950, "precio": 110.00},
    {"min": 0.951, "max": 1.150, "precio": 124.00},  # Ejemplo indicado
    {"min": 1.151, "max": 1.350, "precio": 138.00},
    {"min": 1.351, "max": 1.550, "precio": 152.00},
    {"min": 1.551, "max": 1.800, "precio": 185.00},
    {"min": 1.801, "max": 2.100, "precio": 215.00}
]

# ==============================================================================
# TABLA 2: Opciones y Adicionales por Esmerilado de Cristales (USD)
# ==============================================================================
tabla_esmerilado = {
    "No": 0.00,
    "Si": 25.00
}

# ==============================================================================
# TABLA 3: Catálogo de Accesorios y Herrajes Opciónales (USD)
# ==============================================================================
tabla_accesorios = {
    "Tirador Toallero Brillante": 20.00,
    "Tirador Pomo y Toallero Brillante": 24.00,
    "Tirador Pomo Brillante": 8.00,
    "Perfiles en Color Negro Mate / Dorado": 15.00,
    "Tirador Pomo Negro": 10.00
}

print("✅ Base de datos de 3 Tablas cargada correctamente.")
```

```python
# --- DATOS DEL CLIENTE ---
nombre_cliente = "Eithan Silva"
referencia_proyecto = "Baño Principal"

# --- 1. MEDIDA DEL ANCHO (En metros) ---
ancho_puerta = 1.555  # Ejemplo: 1.30 metro (Caerá en el rango 0.951 - 1.15)

# --- 2. SELECCIÓN DE ESMERILADO (Escribir tal cual la clave de la Tabla 2) ---
opcion_esmerilado_seleccionada = "Si"

# --- 3. SELECCIÓN DE ACCESORIOS (Lista de accesorios elegidos de la Tabla 3) ---
accesorios_seleccionados = [
    "Tirador Pomo Negro"
]
```

```python
# --- 1. EVALUAR TABLA 1: Búsqueda en Rangos de Ancho ---
precio_tabla1 = 0.0
rango_aplicado = "Fuera de rango"

for rango in tabla_rangos_ancho:
    if rango["min"] <= ancho_puerta <= rango["max"]:
        precio_tabla1 = rango["precio"] + 100
        rango_aplicado = f"{rango['min']:.3f} m a {rango['max']:.3f} m"
        break

# Alerta por si el ancho no entra en ningún rango
if precio_tabla1 == 0.0:
    print(f"⚠️ ADVERTENCIA: El ancho ingresado ({ancho_puerta}m) no está dentro de ningún rango de la Tabla 1.")

# --- 2. EVALUAR TABLA 2: Esmerilado ---
precio_tabla2 = tabla_esmerilado.get(opcion_esmerilado_seleccionada, 0.0)

# --- 3. EVALUAR TABLA 3: Suma de Accesorios ---
precio_tabla3 = 0.0
desglose_accesorios = []

for acc in accesorios_seleccionados:
    if acc in tabla_accesorios:
        costo_acc = tabla_accesorios[acc]
        precio_tabla3 += costo_acc
        desglose_accesorios.append((acc, costo_acc))

# --- CÁLCULO TOTAL ---
total_cotizacion = precio_tabla1 + precio_tabla2 + precio_tabla3

# --- GENERACIÓN DEL RESUMEN DETALLADO ---
print("=" * 55)
print("📄 COTIZACIÓN DE PUERTA DE BAÑO CORREDIZA")
print(f"👤 Cliente: {nombre_cliente} | 📍 Ubicación: {referencia_proyecto}")
print("=" * 55)
print(f"📏 Ancho ingresado:       {ancho_puerta:.2f} m")
print(f"📌 Rango asignado:        [{rango_aplicado}]")
print(f"💵 Precio Base (Tabla 1): ${precio_tabla1:.2f}")
print("-" * 55)
print("🎨 Esmerilado (Tabla 2):")
print(f"  • {opcion_esmerilado_seleccionada}: +${precio_tabla2:.2f}")
print("-" * 55)
print("🛠️ Accesorios Seleccionados (Tabla 3):")
if desglose_accesorios:
    for nombre_acc, costo in desglose_accesorios:
        print(f"  • {nombre_acc}: +${costo:.2f}")
else:
    print("  • Ningún accesorio adicional seleccionado.")
print("=" * 55)
print(f"💰 PRECIO TOTAL A COBRAR: ${total_cotizacion:.2f}")
print("=" * 55)
```
